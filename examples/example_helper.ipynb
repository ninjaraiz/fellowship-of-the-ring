## 1. Qué implementación me ha tocado

FRODO reenvía la mayoría de las llamadas a su cuarteto, y `extract_inputs` y
`extract_outputs` las vuelve a declarar como `(*args, **kwargs)`. Resultado:
`help(db.extract_inputs)` te enseña una firma vacía y la prosa genérica de
FRODO, mientras que la firma que tienes que escribir vive en el reader de tu
formato — y cuál sea depende del objeto que tengas en la mano.

````python
from FotR import FRODO

db = FRODO(root_dir='/data/TIFON', format='CODA', name='tifon')

help(db.extract_inputs)      # extract_inputs(self, *args, **kwargs) → inútil
````

coincidencia, y solo para leer su firma.
