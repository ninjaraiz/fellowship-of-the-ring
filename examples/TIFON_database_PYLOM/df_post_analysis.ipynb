```python
import pandas as pd, numpy as np

df_post_basic = pd.read_csv(
    '/home/m.jaraiz/Documentos/DATASETS/data_TIFON/rans3_basic/metadata/df_post.csv',
    sep=',',
    index_col=0
)

df_post_transonic = pd.read_csv(
    '/home/m.jaraiz/Documentos/DATASETS/data_TIFON/rans3_transonic_1/metadata/df_post.csv',
    sep=',',
    index_col=0
)
```

```python
common_titles = ['aoa', 'mach']
```

```python
#comprobar los valores repetidos en las columnas de common_titles con isin
for title in common_titles:
    repeated_values = df_post_transonic[title].isin(df_post_basic[title])
    if repeated_values.any():
        print(f'Valores repetidos en la columna "{title}":')
        print(df_post_transonic[repeated_values][title])
```
